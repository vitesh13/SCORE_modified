```python
import base64
import zlib
import struct
import numpy as np

# Paste the full base64 string from the DataArray here
data_b64 = ""

# Step 1: Decode base64
raw = base64.b64decode(data_b64)

# Step 2: Read VTK binary header (UInt32 sizes)
# First 4 bytes = number of blocks (usually 1 for simple arrays)
num_blocks = struct.unpack("I", raw[:4])[0]

# Next 4 bytes = uncompressed size
uncompressed_size = struct.unpack("I", raw[4:8])[0]

# Next 4 bytes = compressed size
compressed_size = struct.unpack("I", raw[8:12])[0]

# Step 3: Extract compressed payload
compressed_data = raw[12:12 + compressed_size]

# Step 4: Decompress with zlib
decompressed = zlib.decompress(base64.b64decode(data_b64))  # compressed_data)

# Step 5: Convert to numpy array (Int64 in your case)
array = np.frombuffer(decompressed, dtype=np.int64)

print(array)
```
