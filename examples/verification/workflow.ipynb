```python
import os
import shutil

import numpy as np
from pint import Quantity
from score_config import ScoreConfig
from score_config_utils import ureg
from jupyterlab_h5web import H5Web
import subprocess

cwd_path = os.getcwd()
print(cwd_path)
```

## Create a configuration file for SCORE

This generates a NeXus/HDF5 file that matches the `NXmicrostructure_score_config.nxdl.xml` contributed application definition.

```python
print("Generating a NeXus/HDF5 configuration file for examples/verification")
nxs = ScoreConfig()
nxs.deformation()
nxs.nucleation()
nxs.numerics()
nxs.get_config()
nxs.write_config_hfive_file()
```

```python
# optionally inspect NeXus/HDF5 file
# H5Web(f"SCORE.Config.SimID.{nxs.sim_id}.nxs")
```

## Optionally run the simulation using this configuration file

```python
cmd = f"mpiexec -n 1 ./../../build/score {nxs.sim_id} {nxs.hdf5_file_name}"
p = subprocess.Popen(cmd, shell=True)
while p.poll() == None:
    p.poll()                                                                                                                                                                                                                                     
if p.poll() == 0:                                                                                                       
   print(f'Recrystallization simulation completed')
```
